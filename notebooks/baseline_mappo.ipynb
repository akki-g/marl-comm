```python
from benchmarl.algorithms import MappoConfig
from benchmarl.experiment import Experiment, ExperimentConfig
from benchmarl.environments import VmasTask
from benchmarl.models import MlpConfig


experiment_conf = ExperimentConfig.get_from_yaml()
mappo_conf = MappoConfig.get_from_yaml()
simple_spread_conf = VmasTask.SIMPLE_SPREAD.get_from_yaml()
model_conf = MlpConfig.get_from_yaml()
critic_model_config = MlpConfig.get_from_yaml()

experiment_config.max_n_frames = 12_000
experiment_config.loggers = []

experiment = Experiment(
    task = simple_spread_conf,
    algorithm_config=mappo_conf,
    model_config=model_conf,
    seed=0,
    config=experiment_config,,
    critic_model_config=critic_model_config,
)   

experiment.run()
```
